Histogram to understand the distribution

```python
import warnings
warnings.filterwarnings("ignore")
for i in df.select_dtypes(include = "number").columns:
  sns.histplot(data = df, x = i)
  plt.show()
```

Boxplot to identify outliers

```python
import warnings
warnings.filterwarnings("ignore")
for i in df.select_dtypes(include = "number").columns:
  sns.boxplot(data = df, x = i)
  plt.show()
```
